```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data["Sex"]=le.fit_transform(data["Sex"])
data["ChestPainType"]=le.fit_transform(data["ChestPainType"])
data["RestingECG"]=le.fit_transform(data["RestingECG"])
data["ExerciseAngina"]=le.fit_transform(data["ExerciseAngina"])
data["ST_Slope"]=le.fit_transform(data["ST_Slope"])
```

```python
X=data.drop("HeartDisease",axis=1)
y=data["HeartDisease"]
```

```python
sns.heatmap(X.corr(),annot=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
y_pre=model.predict(X_test)
```

```python
acc=accuracy_score(y_test,y_pre)
acc
```

```python
print(classification_report(y_test,y_pre))
```

```python
confusion_matrix(y_test,y_pre)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
model_knn=KNeighborsClassifier(n_neighbors=19)
model_knn.fit(X_train_scaled,y_train)
```

```python
y_knn_pre=model_knn.predict(X_test_scaled)
```

```python
acc=accuracy_score(y_test,y_knn_pre)
acc
```

```python
print(classification_report(y_test,y_knn_pre))
```

```python
model_dt=DecisionTreeClassifier()
model_dt.fit(X_train_scaled,y_train)
y_dt_pre=model_dt.predict(X_test_scaled)
acc=accuracy_score(y_test,y_dt_pre)
acc
```

```python
print(classification_report(y_test,y_dt_pre))
```

```python
confusion_matrix(y_test,y_dt_pre)
```

```python
model_nb=GaussianNB()
model_nb.fit(X_train,y_train)
y_nb_pre=model_nb.predict(X_test)
acc=accuracy_score(y_test,y_nb_pre)
acc
```

```python
print(classification_report(y_test,y_nb_pre))
```

```python
confusion_matrix(y_test,y_nb_pre)
```

```python
model_svm=SVC()
model_svm.fit(X_train,y_train)
y_svm_pre=model_svm.predict(X_test)
acc=accuracy_score(y_test,y_svm_pre)
acc
```

```python
print(classification_report(y_test,y_svm_pre))
```

```python
confusion_matrix(y_test,y_svm_pre)
```
